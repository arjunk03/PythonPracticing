```python
import functools

def add_st(func):
    @functools.wraps(func)
    def _add_st(*args, **kw):
        print("before", args, kw)
        ot = func(*args, **kw)
        print("after")
        return ot

    return _add_st

@add_st
def test(a, d, b=2):
    print("testestsetest", a , b, d)

    return 6
    

ot = test(a=3, b=4, d=5)
print(ot)
help(test)
```

```python
import functools

def memoize(func):
    func.cache = dict()
    
    @functools.wraps(func)
    def _memoize(*args, **kw):
        print("before", args, kw)
        total_args = list(args+tuple(kw.values()))
        total_args = tuple(sorted(total_args))
        if total_args in func.cache:
            print("memmo inside")
            return func.cache[total_args]
            
        func.cache[total_args] = func(*args, **kw)
        print("after")
        return func.cache[total_args]

    return _memoize

@memoize
def test(a, d, b=2):
    print("testestsetest", a , b, d)

    return 6
    

ot = test(a=3, b=4, d=5)

ot = test(a=3, b=5, d=5)
ot = test(a=3, b=5, d=5)

ot = test(a=3, b=4, d=5)
print(ot)
help(test)
print(test.__wrapped__.cache)
```
